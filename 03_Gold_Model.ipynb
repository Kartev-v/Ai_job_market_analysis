```python
from pyspark.sql.functions import (
    col,
    count,
    avg,
    max,
    min,
    current_timestamp,
    explode
)
```

```python
spark.sql("USE CATALOG skitkartev")

spark.sql("""
CREATE SCHEMA IF NOT EXISTS gold
""")
```

```python
silver_df = spark.table("skitkartev.silver.jobs")
```

```python
jobs_fact = (
    silver_df
    .withColumn(
        "GoldProcessedTimestamp",
        current_timestamp()
    )
)
```

```python
jobs_fact.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("skitkartev.gold.jobs_fact")
```

```python
skill_df = (
    silver_df
    .select(
        explode("RequiredSkills").alias("Skill")
    )
)
```

```python
skill_demand = (
    skill_df
    .groupBy("Skill")
    .agg(
        count("*").alias("TotalJobs")
    )
    .orderBy(col("TotalJobs").desc())
)
```

```python
skill_demand.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("skitkartev.gold.skill_demand")
```

```python
company_hiring = (
    silver_df
    .groupBy("CompanyName")
    .agg(
        count("*").alias("TotalJobs")
    )
    .orderBy(col("TotalJobs").desc())
)
```

```python
company_hiring.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("skitkartev.gold.company_hiring")
```

```python
location_hiring = (
    silver_df
    .groupBy(
        "JobLocationCountry",
        "JobLocationState",
        "JobLocationCity"
    )
    .agg(
        count("*").alias("TotalJobs")
    )
)
```

```python
location_hiring.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("skitkartev.gold.location_hiring")
```

```python
salary_summary = (
    silver_df
    .groupBy("ExperienceLevel")
    .agg(
        avg("SalaryMin").alias("AverageSalaryMin"),
        avg("SalaryMax").alias("AverageSalaryMax"),
        min("SalaryMin").alias("MinimumSalary"),
        max("SalaryMax").alias("MaximumSalary")
    )
)
```

```python
salary_summary.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("skitkartev.gold.salary_summary")
```

```python
from pyspark.sql.functions import sum as spark_sum

tech_stack_demand = silver_df.agg(
    spark_sum(col("AzureRequired").cast("int")).alias("AzureJobs"),
    spark_sum(col("AWSRequired").cast("int")).alias("AWSJobs"),
    spark_sum(col("GCPRequired").cast("int")).alias("GCPJobs"),
    spark_sum(col("PythonRequired").cast("int")).alias("PythonJobs"),
    spark_sum(col("SQLRequired").cast("int")).alias("SQLJobs"),
    spark_sum(col("SparkRequired").cast("int")).alias("SparkJobs"),
    spark_sum(col("DatabricksRequired").cast("int")).alias("DatabricksJobs"),
    spark_sum(col("AIRequired").cast("int")).alias("AIJobs"),
    spark_sum(col("GenerativeAIRequired").cast("int")).alias("GenerativeAIJobs")
).withColumn(
    "GoldProcessedTimestamp",
    current_timestamp()
)
```

```python
tech_stack_demand.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("skitkartev.gold.tech_stack_demand")
```
